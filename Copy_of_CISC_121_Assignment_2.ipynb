```python
# Final working solution

MAX  = 0x7FFFFFFF #2 147 483 647
MASK = 0xFFFFFFFF #4 294 967 295

class Solution(object):
    def getSum(self, a, b):

        a &= MASK # limit to 32 bits
        b &= MASK # limit to 32 bits

        #while we still have a carry
        while b != 0:
            bit_sum = (a ^ b) & MASK  # add each bit no carrying
            carry = ((a & b) << 1) & MASK   # carry now
            a = bit_sum
            b =carry #assn carry for next loop

        if a <= MAX:
            return a
        else: # so its negative
            return ~(a ^ MASK) # #32b negative to inta

```
